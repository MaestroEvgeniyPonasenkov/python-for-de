```python
import json, csv
```

```python
def get_purchases():
    purchases = dict()

    with open('purchase_log.txt', 'r', encoding='utf-8') as file:
        next(file)

        for line in file:
            parsed_line = json.loads(line.strip())
            user_id = parsed_line['user_id']
            purchases[user_id] = parsed_line['category']

    return purchases
```

## Задание 1

```python
purchases = get_purchases()
```

```python
i = 0

for user_id, category in purchases.items():
    i += 1
    if i > 10:
        break
    print(f"{user_id} '{category}'")
```

## Задание 2

```python
with (open('visit_log.csv', 'r', encoding='utf-8') as f_visits,
      open('funnel.csv', 'w', encoding='utf-8', newline='') as f_result):
    reader = csv.reader(f_visits)
    next(reader)
    
    writer = csv.writer(f_result)
    writer.writerow(['user_id', 'source', 'category'])

    purchases = get_purchases()

    for user_id, source in reader:
        category = purchases.get(user_id)
        if category is not None:
            writer.writerow([user_id, source, category])
```
